# 02. Price positioning: who actually substitutes?

Tests **H3**. This notebook bounds alpha from above, so it is what defends the
baseline in notebook 01's sensitivity grid.

| | |
|---|---|
| **Statement** | Management claims Bistro serves affordable home style and snacky occasions. If Bistro's price distribution sits well below Zomato's, the two products serve different willingness to pay segments and the substitutable population is structurally small |
| **Test** | Hand sampled menu price distributions and overlap share |
| **Kill condition** | The distributions substantially overlap |

**Frame it as segmentation, not as a price comparison.** Two products at
different price points are not one market with a gap. They are two segments
with different willingness to pay, sorted by self selection. A customer
choosing a Rs 150 snack is not a customer who would otherwise have bought a
Rs 450 dinner. Alpha cannot exceed the share of orders that could plausibly
have gone either way.

That is the argument. The histogram below is the evidence for it.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt

from src import config, panel, metrics, viz

viz.apply_style()
pd.set_option("display.max_columns", 60)
pd.set_option("display.width", 160)

menu = pd.read_csv(config.MENU_SAMPLE)
menu.groupby("platform")["price"].describe()
```

## Price distributions and overlap

```python
fig, ax = plt.subplots()
for plat, colour in [("bistro", viz.PALETTE["qc"]), ("zomato", viz.PALETTE["fd"])]:
    sub = menu[menu["platform"] == plat]["price"]
    ax.hist(sub, bins=20, alpha=0.55, label=f"{plat} (n={len(sub)})", color=colour)
ax.set_title("H3: price distribution by platform")
ax.set_xlabel("Item price, rupees")
ax.set_ylabel("Items")
ax.legend()
viz.caption(fig, "Hand-sampled menu prices. See sampling frame above. "
                 "Not a census; representativeness is unverified.")
viz.save(fig, "h3_price_distributions")
plt.show()
```
