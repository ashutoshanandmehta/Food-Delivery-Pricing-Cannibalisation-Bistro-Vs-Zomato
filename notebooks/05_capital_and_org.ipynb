## H6: where the capital goes

Segment revenue share, store and kitchen additions, and delivery cost growth
against order growth. Even with zero revenue cannibalisation, the two
businesses compete for the same capex, marketing budget and rider capacity.

```python
fig, ax = plt.subplots()
ax.plot(df["quarter"], df["qc_revenue_share"], marker="o", color=viz.PALETTE["qc"])
viz.pct_axis(ax)
ax.set_title("H6: quick commerce share of consolidated revenue")
ax.set_ylabel("Share of consolidated revenue")
plt.xticks(rotation=45, ha="right")
viz.mark_event(ax, config.BISTRO_START_QUARTER, df["quarter"].tolist(), "Bistro launch")
viz.caption(fig, "Revenue share is inflated by Blinkit's inventory-led (1P) "
                 "accounting, which books full order value as revenue. "
                 "Read alongside the NOV-based share, not instead of it.")
viz.save(fig, "h6_revenue_share")
plt.show()
```

**Important caveat to state on the slide.** Revenue share overstates Blinkit's
economic weight because 1P accounting books the full order value while Zomato's
