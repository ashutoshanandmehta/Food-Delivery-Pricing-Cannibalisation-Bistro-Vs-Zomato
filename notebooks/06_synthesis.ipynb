## Limitations

State these plainly. Interviewers probe assumptions harder than conclusions, so
visible limitations are an asset rather than a weakness.

1. No city level or user level data is public, so no true difference in
   differences is possible. All findings here are circumstantial.
2. Bistro is not broken out separately in the filings. Its effect is inferred
   from the quick commerce aggregate.
3. Blinkit's inventory led accounting breaks revenue comparability across
   periods. NOV is used for all growth comparisons.
4. Swiggy is a contaminated control. It operates Bolt and Snacc.
5. Kitchen counts come from press reports rather than filings and are
   approximate.
6. Menu prices are hand sampled, not a census. See notebook 02 for the frame.
7. Roughly ten quarters of data. No statistical power for formal testing. All
   comparisons are descriptive.
8. CM_bistro is built from benchmarks, not disclosure. It is the weakest number
   in the project and the deck must say so on slide 1.
